```python
itens = []

def main():
  print('Bem vindo ao Sistema de Gerenciamento de Lista de compras.')

  while True:
    print('\nMenu:')
    print('1. Adicionar Itens')
    print('2. Visualizar Itens')
    print('3. Marcar Itens como comprado')
    print('4. Remover Itens')
    print('5. Alterar Quantidade')
    print('6. Sair')

    escolha = input('Digite o número que corresponde a opção desejada: ')

    if escolha == '1':
      nome = input('Nome do item a ser adicionado: ')
      quantidade = int(input('Quantidade do item: '))
      adicionar_item(nome, quantidade)

    elif escolha == '2':
      visualizar_item()

    elif escolha == '3':
      visualizar_item()
      try:
        indice = int(input('Digite o número do item que deseja marcar como comprado: ')) - 1
        marcar_como_comprado(indice)
      except ValueError:
        print('Entrada inválida. Por favor, digite um número válido.')

    elif escolha == '4':
      visualizar_item()
      try:
        indice = int(input('Digite o número do item que deseja remover: ')) - 1
        remover_item(indice)
      except ValueError:
        print('Entrada inválida. Por favor, digite um número válido.')
    elif escolha == '5':
      visualizar_item()
      try:
        indice = int(input('Digite o número do item que deseja alterar quantidade: ')) - 1
        alterar_quantidade(indice)
      except ValueError:
        print('Entrada inválida. Por favor, digite um número válido.')


    elif escolha == '6':
      print('Saindo do Sistema')
      break
    else:
      print('Opção inválida. Por favor, escolha uma opção válida')

## --ADICIONAR ITENS
def adicionar_item(nome, quantidade):
  if not nome:
      print('O nome não pode ser vazio.')
      return

  if not nome.replace(' ', '').isalpha():
      print('O nome deve conter somente letras.')
      return

  if not isinstance(quantidade, int):
      print('A quantidade deve ser um número inteiro.')
      return

  if quantidade <= 0:
      print('A quantidade deve ser maior que 0.')
      return

  itens.append({
      'nome': nome,
      'quantidade': quantidade,
      'comprado': False})

  print(f"Item '{nome}' adicionado com sucesso!")

## --VISUALIZAR INTES
def visualizar_item():
  if not itens:
    print('Nenhum item encontrado!')
    return
  print('\nLista de Itens: ')
  for i, item in enumerate(itens):
    status = '[X]' if item['comprado'] else '[ ]'
    print(f'{i + 1}. {status} {item['nome']} - {item['quantidade']}')

## --MARCAR ITENS COMO COMPRADO
def marcar_como_comprado(indice):
  if 0 <= indice <= len(itens):
    itens[indice]['comprado'] = True
    print(f'Item {itens[indice]['nome']} foi marcado como comprado!')
  else:
    print('Índice inválido. Por favor, escolha um índice válido.')

## -- ALTERAR QUATIDADE
def alterar_quantidade(indice):
  if 0 <= indice < len(itens):
    nova_quantidade = int(
        input(f"Digite a nova quantidade para {itens[indice]['nome']}: "))

    if nova_quantidade <= 0:
      print('Quantidade tem que ser maior que 0!')
      return
    elif nova_quantidade == itens[indice]['quantidade']:
      print(f"A quantidade de {itens[indice]['nome']} não foi alterada.")
      return
    elif nova_quantidade < itens[indice]['quantidade']:
      print(f"A nova quantidade de {itens[indice]['nome']} não pode ser menor que a quantidade atual.")
      return
    itens[indice]['quantidade'] = nova_quantidade

    print(f"Quantidade de {itens[indice]['nome']} alterada para {nova_quantidade}!")

## --REMOVE ITENS
def remover_item(indice):
  if 0 <= indice <= len(itens):
    item_removido = itens.pop(indice)
    print(f'Item {item_removido['nome']} foi removida com sucesso!')
```

```python
main()
```
